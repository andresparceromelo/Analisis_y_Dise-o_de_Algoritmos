 PROBLEMA 5: Robo de Casas
 ═══════════════════════════════════════════════════════════════════════════

 Dado un arreglo con el dinero de cada casa. No se pueden robar dos casas
 adyacentes. Retornar el máximo dinero que se puede robar.

 Ejemplo:
   [1, 2, 3, 1]      → 4   (casas 0 y 2)
   [2, 7, 9, 3, 1]   → 12  (casas 0, 2, 4)



### Análisis Detallado: Estrategia de Robo de Casas

**1. Definición del Subproblema**
El problema nos pide maximizar el dinero robado sin tomar dos casas adyacentes. Esto implica que en cada casa `i`, tenemos dos opciones:
- **Robar la casa `i`**: Si la robamos, no podemos haber robado la `i-1`. El botín será lo que hay en la casa `i` más el mejor botín acumulado hasta la casa `i-2`.
- **No robar la casa `i`**: Si la ignoramos, el botín máximo será el mismo que el mejor botín acumulado hasta la casa `i-1`.

**2. Relación de Recurrencia**
Si definimos `dp[i]` como la ganancia máxima al considerar hasta la casa `i`, la transición se define como:
> `dp[i] = max(dp[i-2] + dinero[i], dp[i-1])`

**3. Casos Base**
- Para `n=0`: Ganancia es `0`.
- Para `n=1`: Ganancia es `dinero[0]`.
- Para `n=2`: Ganancia es `max(dinero[0], dinero[1])`.

**4. Estrategia Utilizada: DP Iterativo (Tabulación) con Optimización de Espacio**
El enfoque clásico de Programación Dinámica usaría un arreglo `dp` de tamaño `n`, lo cual requeriría una memoria de $\mathcal{O}(n)$. Sin embargo, al observar la recurrencia, notamos que para calcular la posición actual **solo necesitamos los dos últimos estados** (`i-1` y `i-2`).

Por lo tanto, la solución optimiza de manera inteligente el uso de memoria al reemplazar todo el arreglo por únicamente dos variables escalares:
- `dos_atras` $\rightarrow$ Representa `dp[i-2]`
- `uno_atras` $\rightarrow$ Representa `dp[i-1]`

En cada iteración del ciclo, calculamos la ganancia máxima para la casa actual y luego **desplazamos la ventana** de nuestras variables: `dos_atras` pasa a valer lo que valía `uno_atras`, y `uno_atras` pasa a tener la nueva ganancia actual.

**5. Complejidad Algorítmica**
- **Tiempo:** $\mathcal{O}(n)$ - Realizamos un único recorrido lineal por las casas, realizando operaciones de tiempo constante $\mathcal{O}(1)$ en cada paso.
- **Espacio:** $\mathcal{O}(1)$ - Solo se usan dos variables extra independientemente de la cantidad de casas. Esta es la solución óptima teórica para el problema.

```python
# ─────────────────────────────────────────────
# 5. ROBO DE CASAS (House Robber)
# Estrategia: DP Iterativo con variables escalares (optimización de espacio)
# T: O(n)   |   S: O(1)
# ─────────────────────────────────────────────
def robo_casas(dinero):
    """Máximo dinero sin robar casas adyacentes."""
    n = len(dinero)
    if n == 0:
        return 0
    if n == 1:
        return dinero[0]

    # dos_atras = dp[i-2]: mejor botín hasta dos posiciones atrás
    # uno_atras  = dp[i-1]: mejor botín hasta la posición anterior
    dos_atras = dinero[0]
    uno_atras  = dinero[0] if dinero[0] > dinero[1] else dinero[1]

    for i in range(2, n):
        # Robar casa i: sumar dinero[i] al mejor botín dos casas atrás
        robar_actual   = dos_atras + dinero[i]
        # No robar casa i: conservar el mejor botín previo
        no_robar_actual = uno_atras

        actual    = robar_actual if robar_actual > no_robar_actual else no_robar_actual
        dos_atras = uno_atras
        uno_atras  = actual

    return uno_atras
```
