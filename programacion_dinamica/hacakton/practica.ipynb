```python
#fibonacci

#sin memo
def fib_memont(n):
    if n <= 1:
        return n

    return fib_memont(n-1) + fib_memont(n-2)



#con memo
def fib_memo(n, memo ={}):
    if n in memo:
        return memo[n]

    if n <= 1:
        return n
    
    memo[n] = fib_memo(n-1) + fib_memo(n-2)

    return memo[n]

print(fib_memont(5))
print(fib_memo(4))
```

```python
#stairs

def stairs(n, memo = {}):
    if n in memo:
        return memo[n]
    
    if n <= 2:
        return n
    
    formas_llegar = stairs(n-1) + stairs(n-2)

    memo[n] = formas_llegar

    return memo[n]

stairs(740)
```
