```python
#quiero la potencia de un numero n, con potencia k
def potencia(x, k):
    
    if k == 0:
        return 1

    if k == 1:
        return x

    mid = k//2

    left = potencia (x, mid)
    right = potencia(x, k-mid)
    return left * right

prueba = potencia(2,7)
prueba


    
    

```

```python
#suma elementos de un arreglo

def suma_arr(arr, suma = 0, izq = 0, der=None,):
    if der == None:
        der = len(arr) -1

    if der - izq == 1:
        return suma + arr[der] + arr[izq]

    if izq == der:
        return suma + arr[izq]
    
    medio = (izq+der) // 2

    suma_izq = suma_arr(arr, suma, izq, medio)
    suma_der = suma_arr(arr, suma, medio +1, der)

    return suma_izq + suma_der

print(suma_arr([1,2,3,4,5,10,10,10]))
    


```

```python
# quick_sort que me reciba k y me retorne el k-esimo numero mas pequeño
# se resuelve con quick_select()

def quick_sort(arr, k):

    if len(arr) <= 1:
        return arr

    pivote = arr[0]
    menores = [x for x in arr[1: ] if x <= pivote]
    mayores = [x for x in arr[1: ] if x > pivote]


    if len(menores) > k:

        return quick_sort(menores) + [pivote]


    return quick_sort(menores) + [pivote] + quick_sort(mayores)

    
    

    test = quick_sort([5,10,4,1,3,7])
```

```python
def maximo_rotado_solucion(arr):
    def dyv(arr, izq, der):
        # Caso base: un solo elemento
        if izq == der:
            return arr[izq]

        # Si el subarreglo está ordenado, el máximo es el último
        if arr[izq] <= arr[der]:
            return arr[der]

        medio = (izq + der) // 2

        # El máximo está donde ocurre el quiebre
        # Si la mitad izquierda está ordenada, el quiebre está a la derecha
        if arr[izq] <= arr[medio]:
            return dyv(arr, medio + 1, der) if arr[medio] < arr[medio + 1] \
                else arr[medio] if medio + 1 > der or arr[medio] > arr[medio + 1] \
                else dyv(arr, medio + 1, der)
        else:
            return dyv(arr, izq, medio)

    if not arr:
        return None
    return dyv(arr, 0, len(arr) - 1)
```

```python
#Encontrar el primer elemento que cumple que sea mayor que el que le sigue

def find_index(arr, izq = 0, der = None):
    if der is None:
        der = len(arr) - 1
    
    if izq >= der:
        return -1
    
    mid = (izq+der) // 2

    if mid < der and arr[mid] > arr[mid+1]:
        posible = mid

    else:
        posible = -1

    left = find_index(arr, izq, mid)


    if left != -1:
        return left

    if posible != -1:
        return posible

    return find_index(arr, mid + 1, der)

find_index([9,1,2])

# [1,2,3,7,5,6] → 3
# [1,2,3,4]     → -1
# [9,1,2]       → 0
```
